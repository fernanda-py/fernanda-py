# Método de Euler

Para encontrar solución a algunas ecuaciones diferenciales podemos utilizar el método de Euler. Supongamos que tenemos una ecuación $x(t)$ y la expandimos con Taylor:
$$x(t+h)=x(t)+h \frac{dx}{dt}+ \frac{1}{2}h²\frac{d^2x}{dt²}$$
$$x(t+h)=x(t)+hf(x,t)+O(h²)$$

Para el método de Euler requerimos de: $f(x,t)$, $[a,b]$, N o H en las cuales están determinadas por $h=\frac{b-a}{N}$, $N=\frac{b-a}{h}$

El error en el método de Euler está dado por $O(h²)$, $t_k=a +kh$; $x_k=x(t_k)$ es lineal

Euler nos dice que nos quedaremos con la parte lineal de una interpolación lineal de la serie de Taylor.

Para interpolar dependíamos de que existiera la derivada, así el inconveniente que pedíamos en interpolación está dado. Esto es porque en ecs dif estamos aproximando clase C1 y se nos está dando explícitamente una derivada. Nos conviene regresar a series de Taylor.

Los errores van a depender del tamaño del intervalo al cuadrado

## Ejemplos

```python
import numpy as np
import matplotlib.pyplot as plt

#Intervalo
a = 0
b = 10
#Número de pasos
N = 1000
#Tamaño de paso
h = (b - a) / N
#Paso incial
x0 = 0

f = lambda x,t: -x**3 + np.sin(t)

x_values = [x0]
t_values = np.linspace(a, b, N+1)

# Método de Euler
for t in t_values[:-1]:
    x_th = x0 + h * f(x0, t)
    x_values.append(x_th)
    x0 = x_th

plt.plot(t_values,x_values)
```

```python
import math
a=1
b=1.5
N = 1000
h = (b - a) / N
y0=0
x= np.linspace(a,b,N)

def g(x,y):
      return (np.exp(-x) * (np.tan(x)/x))-y

valores1= [y0]
x_values = np.linspace(a, b, N)

# Método de Euler
for x in x_values[:-1]:
    y_th = y0 + h * g(x, y0)
    valores1.append(y_th)
    y0 = y_th

plt.plot(x_values,valores1,'y')
```

```python
#Ejercicio 3
a=0
b=5
N=1000
h=(b-a)/ N
y0=100
v0=5
g=9.81

def k(t):
      return -g*t+v0

valores2=[y0]
t_valores=np.linspace(0,5,1000)

#Ahora usando método de Euler


for t in t_valores[:-1]:
    y_th = y0 + h * k(t)
    valores2.append(y_th)
    y0 = y_th

plt.plot(t_valores,valores2,'r')
```
