## Cambio de coordenadas

(i) Crea una serie de funciones que permita cambiar coordenadas cartesianas a cordenadas polares, cilindricas y esfericas.

(ii) Genera un menú que se comunique con el usuario para que ingrese las coordenadas a transformar e indique como se hará la transformación (Del sistema origen al final)


```python
import math as mth
import numpy as np

def cartesianas_polares(x,y):
    r = mth.sqrt(x**2 + y**2)
    theta = mth.atan2(y,x)
    polares = [r,theta]
    return polares

def cartesianas_esfericas(x,y,z):
    rho = mth.sqrt(x**2 + y**2 + z**2)
    theta = mth.acos(z/rho)
    phi = mth.atan2(y,x)
    esfericas = [rho,theta,phi]
    return esfericas

def cartesianas_cilindricas(x,y,z):
    rho = mth.sqrt(x**2 + y**2 + z**2)
    theta = mth.atan2(y,x)
    cilindricas = [rho,theta,z]
    return cilindricas

#Generamos el programa que interactuará con el usuario

transformada = input('Dime a qué sistema coordenado quieres cambiar las coordenadas: ')

x = float(input('Dame la coordenada x de tu vector \t'))
y = float(input('Dame la coordenada y de tu vector \t'))
if transformada == 'esfericas' or transformada == 'cilindricas':
    z = float(input('Dame la coordenada z de tu vector \t'))


if transformada == 'polares':
    print(cartesianas_polares(x,y))
elif transformada == 'esfericas':
    print(cartesianas_esfericas(x,y,z))
elif transformada == 'cilindricas':
    print(cartesianas_cilindricas(x,y,z))
    
```

## Matriz Transpuesta

Escribe una función que reciba una matriz y regrese la matriz transpuesta.

```python
def transpuesta(matriz):
    n = len(matriz)
    M_t = np.empty([n,n],float)
    for i in range(n):
        for j in range(n):
            M_t[i,j] = matriz[j,i]
    return M_t

A=np.array([[1,2],[3,4]])
transpuesta(A)
```

## Funciones de onda del oscilador armónico

Las eigenfunciones de onda para el oscilador armónico cuántico son
\begin{equation*}
    \frac{1}{\sqrt{2^n n!}}
    \left( \frac{m\omega}{\pi\hbar}\right)^{1/4}
    e^{-\frac{m\omega x^2}{2\hbar}}H_n\left(\sqrt{\frac{m\omega}{\hbar}}x\right).
\end{equation*}
Usando *matplotlib*, reproduzca la siguiente figura:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import hermite

# Definir el potencial del oscilador armónico cuántico
def potencial(x, m, omega, hbar):
    return 0.5 * m * omega**2 * x**2

# Definimos la función de las eigenfunciones de onda del oscilador armónico cuántico y para esto separamos la función en "partes"
def eigenfuncion_oscilador_armonico(x, n, m, omega, hbar):
    factorial = 1 / np.sqrt(2**n * np.math.factorial(n))
    coeficiente = (m * omega / (np.pi * hbar))**(1/4)
    exponente = np.exp(-m * omega * x**2 / (2 * hbar))
    hermite_pol = hermite(n)(np.sqrt(m * omega / hbar) * x)
    return factorial * coeficiente * exponente * hermite_pol

# Parámetros 
m = 1  # masa
omega = 1  # frecuencia angular
hbar = 1  # constante reducida de Planck

x = np.linspace(-5, 5, 1000)

V = potencial(x, m, omega, hbar)

plt.figure(figsize=(8, 6))
plt.plot(x, V, label='Potencial')

# Graficar cada eigenfunción encima del potencial
for n in range(5):
    eigenfuncion = eigenfuncion_oscilador_armonico(x, n, m, omega, hbar) + (n+1/2)*hbar*omega
    plt.plot(x, eigenfuncion, label=f'Eigenfunción n={n}')

plt.title('Potencial y eigenfunciones de onda para el oscilador armónico cuántico')
plt.xlabel('x')
plt.ylabel('Valor')
plt.legend()
plt.grid(True)
plt.show()
```

## Proyectil con efectos de fricción

Las componentes horizontal y vertical de la posición para un proyectil con efectos de fricción se pueden modelar respectivamente como
\begin{align*}
    x(t) &= \frac{v_0v_t\cos\theta}{g}\left( 1-e^{-gt/v_t}\right) \\
    y(t) &= \frac{v_t}{g}(v_0\sin\theta+v_t)\left( 1-e^{-gt/v_t}\right) - v_tt,
\end{align*}
con $v_t=mg/c$. Usando *matplotlib*, grafique en la misma figura los casos $c=10^{-3},0.2,1,5$. Puede usar los valores $m=1$, $g=10$, $v_0=\sqrt{10}$ y $\theta=\pi/4$. Debería obtener la siguiente figura:

```python
import matplotlib.pyplot as plt
import math as mth
import numpy as np
#Definimos las constantes
c = [1e-3,0.2,1,5]
m = 1
g = 10
v0 = mth.sqrt(10)
theta = mth.pi / 4

colores = ['#C39BD3', '#48C9B0', '#FA8072','#EF91CB']


#Definimos la función a graficar. Será la raíz de la suma del cuadrado de las componentes

def trayectoria(t,c):
    vt = (m*g) / c
    A = (v0*vt*mth.cos(theta))/g
    x = A*(1-np.exp(-(g*t)/vt))
    B = (vt/g)*(v0*mth.sin(theta) + vt)
    y = B*(1-np.exp(-(g*t)/vt)) - vt*t
    return x + y

X = np.linspace(0,1,100)

plt.figure(figsize=(6,4))
for i in range(4):
    plt.plot(X,trayectoria(X,c[i]), color=colores[i])
plt.grid(color = '#D7DBDD',linewidth = 1)
plt.legend(c)
plt.xlabel('Tiempo')
plt.ylabel('Trayectoria')
plt.title('Posición vs. tiempo')
```
